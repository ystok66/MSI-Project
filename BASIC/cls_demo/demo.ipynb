```python
import sys, os
sys.path.insert(0, os.path.dirname(os.path.abspath('__file__')))
from cls_api import CLSDemo, noise_curve, show_prediction

# ── mini-SCAN 训练样本 ──────────────────────────────────
# 名词: 1→BLUE, 2→RED, 3→GREEN, DAX→YELLOW
# 操作符: after(交换), thrice(重复3次), surround(包围)
support = [
    # 单名词
    (['1'],                         ['BLUE']),
    (['3'],                         ['GREEN']),
    (['2'],                         ['RED']),
    (['DAX'],                       ['YELLOW']),
    # after: x after y → [y] [x]
    (['2', 'after', '3'],           ['GREEN', 'RED']),
    (['1', 'after', '2'],           ['RED', 'BLUE']),
    # thrice: x thrice → [x] [x] [x]
    (['2', 'thrice'],               ['RED', 'RED', 'RED']),
    (['1', 'thrice'],               ['BLUE', 'BLUE', 'BLUE']),
    # surround: x surround y → [x] [y] [x]
    (['2', 'surround', '3'],        ['RED', 'GREEN', 'RED']),
    (['3', 'surround', '1'],        ['GREEN', 'BLUE', 'GREEN']),
    # 组合
    (['2', 'thrice', 'after', '3'], ['GREEN', 'RED', 'RED', 'RED']),
    (['3', 'after', '1', 'surround', '2'], ['BLUE', 'RED', 'BLUE', 'GREEN']),
    (['2', 'after', '3', 'thrice'], ['GREEN', 'GREEN', 'GREEN', 'RED']),
    (['3', 'surround', '1', 'after', '2'], ['RED', 'GREEN', 'BLUE', 'GREEN']),
]

# ✏️ 你可以在这里加自己的规则，比如：
# support.append((['DAX', 'twice'], ['YELLOW', 'YELLOW']))

demo = CLSDemo(mode='ast', gauss=True, lab_sigma=10)

demo.study(support)
demo.show_support()
```

```python
demo.predict(['DAX', 'thrice'], expected=['YELLOW', 'YELLOW', 'YELLOW'])
demo.predict(['3', 'after', 'DAX'], expected=['YELLOW', 'GREEN'])
demo.predict(['DAX', 'surround', '2'], expected=['YELLOW', 'RED', 'YELLOW'])
```
